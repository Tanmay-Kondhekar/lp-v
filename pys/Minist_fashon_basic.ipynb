#### 1.

For this 1 st problem statement, download the rice_leaf_diseases dataset, from kaggle

Also if the code takes too much time to run, then change the epochs=5, to epochs=1 or 2, likewise can increase it as well to show more epochs

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load images
train = tf.keras.preprocessing.image_dataset_from_directory(
    "rice_leaf_diseases", image_size=(128,128), batch_size=32)

val = tf.keras.preprocessing.image_dataset_from_directory(
    "rice_leaf_diseases", image_size=(128,128), batch_size=32)

# CNN Model
model = models.Sequential([
    layers.Rescaling(1./255, input_shape=(128,128,3)),
    layers.Conv2D(32,3,activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64,3,activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(len(train.class_names), activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(train, validation_data=val, epochs=5)

# Predict and evaluate
loss, acc = model.evaluate(val)

print("Validation Accuracy:", acc)
print("Validation Loss:", loss)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

# Load data
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
    
# Preprocess
X_train, X_test = X_train/255.0, X_test/255.0

# CNN Model
model = models.Sequential([
    layers.Reshape((28,28,1), input_shape=(28,28)),
    layers.Conv2D(32,3,activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64,3,activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X_train, y_train, epochs=5)

# Predictions
y_pred = model.predict(X_test)
y_pred = y_pred.argmax(axis=1)

# Report
print(classification_report(y_test, y_pred))

# Can skip this if you want, but it shows the confusion matrix!(Read theory once if u are showing it!)
print(confusion_matrix(y_test, y_pred))
```
